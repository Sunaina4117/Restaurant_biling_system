```python

# RESTAURANT BILLING SYSTEM

menu = {
    1: ("Pizza", 200),
    2: ("Burger", 100),
    3: ("Pasta", 150),
    4: ("Coffee", 80)
}

cart = []


# 1. Show Menu
def show_menu():
    print("\n===== RESTAURANT MENU =====")

    for number, item in menu.items():
        print(number, item[0], "- Rs.", item[1])

    print("5. Generate Bill and Exit")


# 2. Order Food
def order_food(choice):
    if choice in menu:
        quantity = int(input("Enter quantity: "))

        if quantity > 0:
            name, price = menu[choice]
            amount = price * quantity

            cart.append((name, quantity, amount))

            print(name, "added successfully!")
            print("Item Total: Rs.", amount)
        else:
            print("Quantity must be greater than 0")

    else:
        print("Invalid item number!")


# 3. Generate Bill
def generate_bill():
    if len(cart) == 0:
        print("No items ordered!")

    else:
        total = 0

        print("\n====== CUSTOMER BILL ======")

        for item in cart:
            print(item[0], "x", item[1], "=", item[2])
            total = total + item[2]

        print("---------------------------")
        print("Total Amount: Rs.", total)


# 4. Main Function
def main():
    while True:
        show_menu()

        choice = int(input("Enter your choice: "))

        if choice in menu:
            order_food(choice)

        elif choice == 5:
            generate_bill()
            print("=========Thank you for visiting=========")
            break

        else:
            print("Invalid choice!")


# Start Project
main()
```
